## 6. Feature Selection

The target variable for prediction is:

LIMIT_BAL

All remaining variables are used as independent variables to predict the credit limit.

```python
#6. Feature Selection
#Define Target Variable
X = df.drop('LIMIT_BAL', axis=1)

y = df['LIMIT_BAL']
```

```python
#Check Features
print(X.columns)
```

##7. Train-Test Split

The dataset is divided into:

- Training Data (80%)
- Testing Data (20%)

The training set is used to train the model, while the testing set evaluates its performance on unseen data.

```python
#7. Train-Test Split
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42)
print(X_train.shape)
print(X_test.shape)
```

We have split the traning and test

## 8. Model Building using Linear Regression

Linear Regression is selected as the baseline machine learning algorithm because:

- It is simple and interpretable.
- It helps understand feature relationships.
- It provides a foundation for comparison with advanced models.

```python
#Create Linear Regression Model
lr = LinearRegression()
#Train Model
lr.fit(X_train, y_train)
```

##9. Model Prediction

After training the model, predictions are generated for the test dataset.

These predicted values represent the estimated credit limits for customers.

```python
# Prediction
y_pred = lr.predict(X_test)
#Actual vs Predicted
comparison = pd.DataFrame({
    'Actual': y_test,
    'Predicted': y_pred
})

comparison.head()
```

## 10. Model Evaluation

The model is evaluated using:

1. Mean Absolute Error (MAE)
2. Root Mean Squared Error (RMSE)
3. R-Squared Score (R²)

```python
#MAE
mae = mean_absolute_error(y_test, y_pred)

print("MAE:", mae)

#RMSE
rmse = np.sqrt(mean_squared_error(y_test, y_pred))

#R² Score
r2 = r2_score(y_test, y_pred)

print("R² Score:", r2)

print("RMSE:", rmse)
```

### Interpretation of Results

R² Score:
Indicates the proportion of variance explained by the model.

MAE:
Represents the average prediction error.

RMSE:
Measures the overall prediction error and penalizes larger mistakes.

A higher R² and lower MAE/RMSE indicate better model performance.

Overall Assessment:

The Linear Regression model demonstrates a moderate predictive capability, explaining about 35% of the variability in credit limits. While the model captures some important relationships within the data, the relatively high MAE and RMSE values indicate room for improvement. Feature engineering, additional relevant variables, or more advanced regression algorithms may help improve prediction accuracy.

```python
#Regression Equation
coefficients = pd.DataFrame({
    'Feature': X.columns,
    'Coefficient': lr.coef_
})

coefficients.sort_values(
    by='Coefficient',
    ascending=False)
```

Linear Regression coefficients help identify which variables have the strongest impact on credit limit prediction.


## 11. Feature Importance Visualization

##13. Residual Analysis

Residuals represent the difference between actual and predicted credit limits.

A good regression model should produce residuals that are approximately normally distributed around zero.

```python
residuals = y_test - y_pred

plt.figure(figsize=(8,5))

sns.histplot(
    residuals,
    kde=True
)

plt.title('Residual Distribution')
```

```python
plt.scatter(y_pred, residuals)
plt.axhline(y=0, color='red', linestyle='--')
plt.xlabel('Predicted Values')
plt.ylabel('Residuals')
plt.title('Residual Plot')
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.scatterplot(x=y_pred, y=residuals)
plt.axhline(y=0, color='red', linestyle='--')
plt.xlabel('Predicted Values')
plt.ylabel('Residuals')
plt.title('Residual Plot')
plt.show()
```

##Interpretation:

Residuals randomly scattered around the red line → model assumptions are reasonably satisfied.
Clear patterns (curve, funnel shape) → possible non-linearity or heteroscedasticity.
Many points far from the line → potential outliers.
